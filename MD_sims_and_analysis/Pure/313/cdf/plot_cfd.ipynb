```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker # Import ticker
from matplotlib.colors import LogNorm

# Read CSV (semicolon-separated)
df = pd.read_csv("cdf_2_rdf[H2r_O1o]_adf[O1r_H2r]-[H2r_O1o]_matrix.csv", sep=";", engine="python")

# Axes
r = df.iloc[:, 0].astype(float).to_numpy()              # distance (pm)
cos_theta = np.array(df.columns[1:], dtype=float)      # cos(angle)

# Heatmap values (NO normalization)
Z = df.iloc[:, 1:].astype(float).to_numpy()

# Plot
plt.figure(figsize=(10, 6))

pcm = plt.pcolormesh(
    r,
    cos_theta,
    Z.T,
    shading="nearest"   # simplest, no bin edge inference
)

# Colorbar with adjustable font size for the label
cbar = plt.colorbar(pcm)
cbar.set_label("occurrence", fontsize=14)
cbar.ax.tick_params(labelsize=14) 

# Labels
plt.ylabel(r"$\angle \mathrm{Or\!-\!Hr\!-\!Oo}$", fontsize=14)
plt.xlabel(r"$r_{\mathrm{Hr\!-\!Oo}}\ \mathrm{(pm)}$", fontsize=14)
plt.xlim((0, 400))

# Ticks
ax = plt.gca()
ax.xaxis.set_major_locator(ticker.MultipleLocator(100))
ax.tick_params(axis='both', which='major', labelsize=14)

plt.tight_layout()
plt.savefig("Pure_2_pentanol_cdf.png",dpi=300)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from matplotlib.colors import LogNorm

# Set global font size
plt.rcParams.update({'font.size': 14})

# Read CSV (semicolon-separated)
# Replace with your actual file path if needed
df = pd.read_csv("cdf_2_rdf[H2r_O1o]_adf[O1r_H2r]-[H2r_O1o]_matrix.csv", sep=";", engine="python")

# Axes
r = df.iloc[:, 0].astype(float).to_numpy()              # distance (pm)
cos_theta = np.array(df.columns[1:], dtype=float)       # cos(angle)

# Heatmap values
Z = df.iloc[:, 1:].astype(float).to_numpy()

# Plot
plt.figure(figsize=(10, 6))

pcm = plt.pcolormesh(
    r,
    cos_theta,
    Z.T,
    norm=LogNorm(),     # Added LogNorm back since it was imported
    shading="nearest"
)

# Colorbar with adjustable font size for the label
cbar = plt.colorbar(pcm)
cbar.set_label("Value", fontsize=14)
cbar.ax.tick_params(labelsize=14) 

# Labels
plt.ylabel(r"$\angle \mathrm{Or\!-\!Hr\!-\!Oo})$", fontsize=14)
plt.xlabel(r"$r_{\mathrm{Hr\!-\!Oo}}\ \mathrm{(pm)}$", fontsize=14)
plt.xlim((0, 400))

# Ticks
ax = plt.gca()
ax.xaxis.set_major_locator(ticker.MultipleLocator(100))
ax.tick_params(axis='both', which='major', labelsize=14)

plt.tight_layout()
plt.show()
```
